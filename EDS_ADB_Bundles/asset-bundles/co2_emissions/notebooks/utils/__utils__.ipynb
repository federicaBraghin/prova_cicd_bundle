```python
import time

def get_config():
    """Legge i parametri del job. I valori dopo la virgola sono i default,
    usati quando lanci il notebook a mano (senza job)."""
    dbutils.widgets.text("catalog", "workspace")
    dbutils.widgets.text("percorso_raw", "/Volumes/workspace/raw/co2_emissions/co2_emissions_canada.csv")
    return {
        "catalog": dbutils.widgets.get("catalog"),
        "percorso_raw": dbutils.widgets.get("percorso_raw"),
    }

def run_query_send_logs(steps_list):
    """Esegue gli step in ordine e scrive i log."""
    for step in steps_list:
        inizio = time.time()
        print(f"Inizio step: {step['name']}")
        if callable(step["query"]):
            step["query"]()             # step Python
        else:
            spark.sql(step["query"])    # step SQL
        print(f"Fine step: {step['name']} in {round(time.time() - inizio, 1)}s")
```
